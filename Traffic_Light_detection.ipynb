# Train model

```python
def train_yolo(
    model_weights: str,
    data_yaml: str,
    epochs: int = 80,
    imgsz: int = 640,
    batch: int | None = None,
    project: str = "runs/train",
    name: str = "exp",
    **kwargs
):
    model = YOLO(model_weights)

    train_args = dict(
        data=data_yaml,
        epochs=epochs,
        imgsz=imgsz,
        project=project,
        name=name,
        **kwargs
    )

    # ✅ Only include batch if provided
    if batch is not None:
        train_args["batch"] = int(batch)

    model.train(**train_args)
    return model
```

```python
train_yolo(
    model_weights="yolov8l.pt",
    data_yaml=f"{dataset_dir}/data.yaml",
    epochs=80,
    imgsz=640
)
```
